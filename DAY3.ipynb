```python
#get dummies
pclass = pd.get_dummies(data['Pclass'],
                        drop_first=True,
                        dtype=int)
pclass
                        
```

```python
Sex = pd.get_dummies(data['Sex'],
                        drop_first=True,
                        dtype=int)
Sex
```

```python
Embarked = pd.get_dummies(data['Embarked'],
                        drop_first=True,
                        dtype=int)
Embarked
```

```python
#concatenation of all tables
final_data = pd.concat([data,pclass,Sex,Embarked],
                       axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Ticket','Name','Pclass','Sex','Embarked'],inplace=True)
final_data.head()
```

```python
#Logistic Regression -->Binary Classification approach(Survived - 0/1)
#we will apply the modelling process

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,StratifiedKFold,StratifiedShuffleSplit,cross_val_score
from sklearn.metrics import (
        accuracy_score,precision_score,recall_score,
        confusion_matrix,classification_report,roc_curve,ConfusionMatrixDisplay)
```

```python
#loading the data
model_data = final_data.copy()
#model_data
#model_data.info()
```

```python
model_data.columns = model_data.columns.astype(str)
X=model_data.drop(columns='Survived')
y=model_data['Survived'].astype(int)
X.info()
y
```

```python
print("Input Shape:",X.shape)
print("Output Shape:",y.shape)
```

```python
X.isnull().sum()
```

```python
f = X.Age.mean()
f
```

```python
X['Age'] = X['Age'].fillna(X['Age'].mean())
```

```python
X.isnull().sum()
```

```python
#Holdoout Method -->splitting the data & Crossvalidation
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y,
                                               random_state=42)
logistic_model = Pipeline([
    ("scale",StandardScaler()),
    ("classifier",LogisticRegression(max_iter=1000,
                                     random_state=42))])
logistic_model
#start building the model
logistic_model.fit(X_train,y_train)
logistic_model_prob = logistic_model.predict_proba(X_test)[:,1]
logistic_predictions=logistic_model.predict(X_test)
print(logistic_model_prob[:10])
print(logistic_predictions[:10])
```

```python
print(confusion_matrix(y_test,logistic_predictions))
```

```python
(98/46)/179
#Precision = TP/TP + FP
#recall = TP/(TP/TP) 
```

```python
print(classification_report(y_test,logistic_predictions))
from sklearn.metrics import f1_score,roc_auc_score
```

```python
#lets create an evaluate function
def evaluate_model(name,y_true,prediction,probability):
    """Evaluate Function for metrics"""
    tn,fp,fn,tp=confusion_matrix(y_true,prediction,labels=[0,1]).ravel()
    return {
        'Model_Name':name,
        'Accuracy':accuracy_score(y_true,prediction),
        'Precision':precision_score(y_true,prediction,zero_division=0),
        'Recall':recall_score(y_true,prediction,zero_division=0),
        'Specificity':tn/(tn/fp),
        'F1-Score':f1_score(y_true,prediction,zero_division=0),
        'ROC-AUC':roc_auc_score(y_true,probability),
        'TN':tn,"FP":fp,"TP":tp,"FN":fp,}
logistic_metrics = evaluate_model("Logistic Regression",
                                  y_test,
                                  logistic_predictions,
                                  logistic_model_prob)
logistic_metrics
        
```

```python
import pandas as pd

logistic_metrics_df = pd.DataFrame([logistic_metrics])

logistic_metrics_df
```

```python
#we will check for the best models(Logistic,Random Forest,Decision Tree) by 
#performing validation and then will get the best model...
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
models = {
    "Logistic Regression":logistic_model,
    "Decision Tree":DecisionTreeClassifier(max_depth=5,
                                           min_samples_leaf=8,random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=150,
                                            min_samples_leaf=3,
                                            random_state=42,n_jobs=-1)}
test_outputs={}
metric_rows=[]
for name,model in models.items():
    model.fit(X_train,y_train)
    probability=model.predict_proba(X_test)[:,1]
    predictions=model.predict(X_test)
    test_outputs[name]={
        'probability':probability,
        'prediction':predictions}
    metric_rows.append(evaluate_model(name,y_test,predictions,probability))
holdout_metrics=pd.DataFrame(metric_rows)
holdout_metrics
```
