```python
data['Embarked']=data['Embarked'].fillna('S')
```

```python
data.isnull().sum()
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
#get dummies
pclass = pd.get_dummies(data['Pclass'],
                        drop_first=True,
                        dtype=int)
pclass
                        
```

```python
Sex = pd.get_dummies(data['Sex'],
                        drop_first=True,
                        dtype=int)
Sex
```

```python
Embarked = pd.get_dummies(data['Embarked'],
                        drop_first=True,
                        dtype=int)
Embarked
```

```python
#concatenation of all tables
final_data = pd.concat([data,pclass,Sex,Embarked],
                       axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Ticket','Name','Pclass','Sex','Embarked'],inplace=True)
final_data.head()
```
